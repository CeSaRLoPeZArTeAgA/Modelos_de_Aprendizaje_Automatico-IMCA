#CLASE 1 - ANALISIS DE SENTIMIENTO



```python
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer
```

#CUENTA LAS PALABRAS QUE SE REPITE - NO HACE

```python
text=['Hello my name is Victor',
      'victor this is my python notebook',
      'Victpr is trying to create a big dataset',
      'features of Count Vectorize',
      'Use of different words']
coun_vect=CountVectorizer()#todo pasa a minuscula por defecto
count_matrix=coun_vect.fit_transform(text)#creamos una matriz dispersa
count_array=count_matrix.toarray()
df=pd.DataFrame(data=count_array,columns=coun_vect.get_feature_names_out())
print(df)
```

#LOWER CASE

```python
text=['Hello my name is Victor',
      'victor this is my python notebook']
coun_vect=CountVectorizer(lowercase=False)#distingue entre mayuscula y minuscula
count_matrix=coun_vect.fit_transform(text)#creamos una matriz dispersa
count_array=count_matrix.toarray()
df=pd.DataFrame(data=count_array,columns=coun_vect.get_feature_names_out())
print(df)
```

#Stop Word

```python
text=['Hello my name is Victor',
      'victor this is my python notebook',
      'Victpr is trying to create a big dataset',
      'features of Count Vectorize',
      'Use of different words']
coun_vect=CountVectorizer(stop_words=['is','to','my'])#excluye esta lista de palabras
count_matrix=coun_vect.fit_transform(text)#creamos una matriz dispersa
count_array=count_matrix.toarray()
df=pd.DataFrame(data=count_array,columns=coun_vect.get_feature_names_out())
print(df)
```

#DEFAULT STOP WORDS

```python
text=['Hello my name is Victor',
      'victor this is my python notebook',
      'Victpr is trying to create a big dataset',
      'features of Count Vectorize',
      'Use of different words']
coun_vect=CountVectorizer(stop_words='english')#quita todas la palabras no relevantes del idioma ingles
count_matrix=coun_vect.fit_transform(text)#creamos una matriz dispersa
count_array=count_matrix.toarray()
df=pd.DataFrame(data=count_array,columns=coun_vect.get_feature_names_out())
print(df)
```

#PALABRAS UNICAS MAS FRECUENTES
